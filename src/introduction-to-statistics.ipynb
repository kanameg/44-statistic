```python
import numpy as np
import pandas as pd
```

```python
# データの平均値を求める
t = np.array([400, 1500, 200, 100, 2000, 5000, 300])
print(np.mean(t))
```

```python
# 23-40, 40-55, 55-70, 70-85, 85-100で階級値を作成する
class_interval = [
    pd.Interval(23, 40, closed='right'), pd.Interval(40, 55, closed='right'),
    pd.Interval(55, 70, closed='right'), pd.Interval(70, 85, closed='right'), 
    pd.Interval(85, 100, closed='right')
]
```

```python
# 各階級の度数を作成する
class_frequency = [6, 16, 17, 19, 7]
```

```python
# 度数分布表を作成する
frequency_distribution_df = pd.DataFrame({'class': class_interval, 'frequency': class_frequency})
print(frequency_distribution_df)
```

```python
# 各階級の階級値を求める
frequency_distribution_df['midpoint'] = frequency_distribution_df['class'].apply(lambda x: (x.left + x.right) / 2)
print(frequency_distribution_df)
```

```python
# 度数分布表の平均値を求める
mean = (frequency_distribution_df['midpoint'] * frequency_distribution_df['frequency']).sum()
mean /= frequency_distribution_df['frequency'].sum()
print(mean)
```

```python
# 度数分布表を作成する
class_value = [40, 60, 80, 100]
class_frequency = [10, 20, 15, 5]

frequency_distribution_df = pd.DataFrame({'class': class_value, 'frequency': class_frequency})
print(frequency_distribution_df)
```

```python
# 度数分布表から平均値を求める
mean = (frequency_distribution_df['class'] * frequency_distribution_df['frequency']).sum()
mean /= frequency_distribution_df['frequency'].sum()
print(mean)
```
